### Exercício 1 (Aula 01): Análise de Produção de Motores, Temperatura e Defeitos
**Objetivo:** Investigar se o aumento da temperatura operacional está correlacionado ao aumento de defeitos.

```python
import pandas as pd
import numpy as np

# 1. Criação do conjunto de dados dos motores industriais
dados_motores = {
    'produto': ['Motor A', 'Motor B', 'Motor A', 'Motor C', 'Motor B'],
    'producao': [100, 120, 110, 90, 130],
    'defeitos': [5, 12, 6, 3, 15],
    'temperatura': [70, 82, 73, 68, 85]
}

df_motores = pd.DataFrame(dados_motores)
print('--- Visualização Inicial dos Dados ---')
display(df_motores)
```

```python
# 2. Selecionando colunas específicas
print('Coluna isolada de temperatura:')
print(df_motores['temperatura'])

print('\nSubconjunto produto e temperatura:')
display(df_motores[['produto', 'temperatura']])
```

```python
# 3. Filtros condicionais
print('Registros com temperatura acima de 75 °C:')
display(df_motores[df_motores['temperatura'] > 75])

print('\nRegistros com temperatura > 75 °C E mais de 10 defeitos:')
display(df_motores[(df_motores['temperatura'] > 75) & (df_motores['defeitos'] > 10)])
```

```python
# 4. Métricas estatísticas básicas
print(f'Média da temperatura: {df_motores["temperatura"].mean():.2f} °C')
print(f'Média da produção: {df_motores["producao"].mean():.2f} unidades')
print(f'Temperatura máxima: {df_motores["temperatura"].max()} °C | Mínima: {df_motores["temperatura"].min()} °C')

# Ordenação decrescente por temperatura
print('\nOrdenado da maior para a menor temperatura:')
display(df_motores.sort_values(by='temperatura', ascending=False))
```

```python
# 5. Agrupamentos (groupby) e agregação
print('Produção média por produto:')
print(df_motores.groupby('produto')['producao'].mean())

print('\nDefeitos médios por produto:')
print(df_motores.groupby('produto')['defeitos'].mean())

print('\nResumo agregado completo:')
display(df_motores.groupby('produto').agg({
    'producao': 'mean',
    'defeitos': ['sum', 'mean'],
    'temperatura': 'mean'
}))
```

```python
# 6. Engenharia de atributos: Taxa percentual de defeitos
df_motores['taxa_defeitos'] = (df_motores['defeitos'] / df_motores['producao']) * 100
print('DataFrame com taxa de defeitos:')
display(df_motores)

# 7. Matriz de correlação de Pearson
matriz_corr = df_motores[['producao', 'defeitos', 'temperatura']].corr()
print('\nMatriz de Correlação:')
display(matriz_corr)

corr_temp_def = matriz_corr.loc['defeitos', 'temperatura']
print(f'\nConclusão: Correlação fortíssima entre Temperatura e Defeitos: r = {corr_temp_def:.4f}')
print('Isso comprova que o aumento de temperatura está fortemente associado ao aumento de peças defeituosas.')
```

### Exercício 2 (Aula 01 / Exercício 1 da Aula 02): Eficiência Produtiva das Máquinas A-E
A gestão deseja comparar 5 máquinas industriais quanto à produção, consumo elétrico e eficiência.

```python
# Criação do DataFrame de Máquinas
df_maquinas = pd.DataFrame({
    'maquina': ['A', 'B', 'C', 'D', 'E'],
    'producao': [1250, 1380, 1190, 1450, 1320],
    'energia': [420, 465, 390, 510, 405]
})

# a) Máquina que produziu mais
idx_max_prod = df_maquinas['producao'].idxmax()
maq_max_prod = df_maquinas.loc[idx_max_prod]
print(f"a) Máquina que produziu mais: Máquina {maq_max_prod['maquina']} ({maq_max_prod['producao']} peças)")

# b) Máquina que menos consumiu energia
idx_min_energ = df_maquinas['energia'].idxmin()
maq_min_energ = df_maquinas.loc[idx_min_energ]
print(f"b) Máquina com menor consumo de energia: Máquina {maq_min_energ['maquina']} ({maq_min_energ['energia']} kWh)")

# c) Produção média das máquinas
print(f"c) Produção média: {df_maquinas['producao'].mean():.2f} peças")

# d) Consumo médio de energia
print(f"d) Consumo médio de energia: {df_maquinas['energia'].mean():.2f} kWh")

# e) Ordenação da maior para a menor produção
print("\ne) Máquinas ordenadas por produção (decrescente):")
display(df_maquinas.sort_values(by='producao', ascending=False))

# f) Criação da coluna de eficiência (produção / energia)
df_maquinas['eficiencia'] = df_maquinas['producao'] / df_maquinas['energia']
print("\nf) DataFrame com coluna de eficiência calculada:")
display(df_maquinas)

# g) Máquina com melhor relação produção / energia
idx_max_efic = df_maquinas['eficiencia'].idxmax()
maq_max_efic = df_maquinas.loc[idx_max_efic]
print(f"g) Máquina mais eficiente: Máquina {maq_max_efic['maquina']} com {maq_max_efic['eficiencia']:.4f} peças por kWh")
```

### Exemplo 2 (Aula 02): Desempenho e Qualidade de Produção por Turno
Base: `producao_por_turno.csv` (100 registros com dados de turnos Manhã, Tarde e Noite).

```python
df_turno = pd.read_csv('producao_por_turno.csv')
print('Shape:', df_turno.shape)
display(df_turno.head())

# a) Quantidade total de peças produzidas no período analisado
total_pecas = df_turno['pecas_produzidas'].sum()
print(f'a) Total de peças produzidas no período: {total_pecas:,} peças')

# b) Turno com maior produção total
prod_por_turno = df_turno.groupby('turno')['pecas_produzidas'].sum()
print(f'\nb) Produção total por turno:\n{prod_por_turno}')
print(f"   -> Turno com maior produção total: {prod_por_turno.idxmax()} ({prod_por_turno.max():,} peças)")

# c) Quantidade média de peças por registro em cada turno
media_por_turno = df_turno.groupby('turno')['pecas_produzidas'].mean()
print(f'\nc) Produção média por registro:\n{media_por_turno}')

# d) Turno com maior percentual de peças defeituosas
resumo_turno = df_turno.groupby('turno')[['pecas_produzidas', 'pecas_defeituosas']].sum()
resumo_turno['pct_defeitos'] = (resumo_turno['pecas_defeituosas'] / resumo_turno['pecas_produzidas']) * 100
print(f'\nd) Percentual de peças defeituosas por turno:\n{resumo_turno}')
print(f"   -> Turno com maior percentual de defeitos: {resumo_turno['pct_defeitos'].idxmax()} ({resumo_turno['pct_defeitos'].max():.2f}%)")

# e) Correlação entre peças produzidas e peças defeituosas
corr_pecas_def = df_turno[['pecas_produzidas', 'pecas_defeituosas']].corr().loc['pecas_produzidas', 'pecas_defeituosas']
print(f'\ne) Correlação linear de Pearson (produção x defeitos): r = {corr_pecas_def:.4f}')
print('   -> Existe correlação positiva moderada (0,37): turnos com maior volume de produção tendem a ter um número ligeiramente maior de peças defeituosas em valor absoluto.')
```

### Exercício 2 (Aula 02): Análise Operacional de Transporte e Logística
Base: `dados_viagens.csv` (100 registros diários de viagens e quilometragem percorrida).

```python
df_viagens = pd.read_csv('dados_viagens.csv')
display(df_viagens.head())

# a) Total de viagens realizadas no período analisado
total_viagens = df_viagens['quantidade_viagens'].sum()
print(f'a) Total de viagens realizadas: {total_viagens:,} viagens')

# b) Dia com maior quantidade de viagens
dia_pico = df_viagens.loc[df_viagens['quantidade_viagens'].idxmax()]
print(f"b) Dia com maior quantidade de viagens: {dia_pico['data']} ({dia_pico['dia_semana']}) com {dia_pico['quantidade_viagens']} viagens")

# c) Média diária de viagens realizadas
media_diaria = df_viagens['quantidade_viagens'].mean()
print(f'c) Média diária de viagens: {media_diaria:.2f} viagens/dia')

# d) Mês em que foi registrada a maior quantidade total de viagens
df_viagens['mes'] = pd.to_datetime(df_viagens['data']).dt.to_period('M').astype(str)
viagens_mes = df_viagens.groupby('mes')['quantidade_viagens'].sum()
print(f'\nd) Viagens por mês:\n{viagens_mes}')
print(f"   -> Mês com maior volume de viagens: {viagens_mes.idxmax()} ({viagens_mes.max():,} viagens)")

# e) Correlação entre quantidade de viagens e distância total percorrida
corr_viag_dist = df_viagens[['quantidade_viagens', 'distancia_total_km']].corr().loc['quantidade_viagens', 'distancia_total_km']
print(f'\ne) Correlação entre viagens e distância total percorrida: r = {corr_viag_dist:.4f}')
print('   -> Existe fortíssima correlação positiva (0,87): dias com mais viagens demandam proporcionalmente maior quilometragem da frota.')
```

### Exercício 3 (Aula 02): Desempenho Comercial do Produto A
Base: `dados_vendas.csv` (100 registros mensais com unidades vendidas, preço unitário e faturamento).

```python
df_vendas = pd.read_csv('dados_vendas.csv')
display(df_vendas.head())

# a) Quantidade total de unidades vendidas durante todo o período analisado
total_unidades = df_vendas['unidades_vendidas'].sum()
print(f'a) Total de unidades vendidas no período: {total_unidades:,} unidades')

# b) Mês que apresentou o maior faturamento
mes_recorde = df_vendas.loc[df_vendas['faturamento_rs'].idxmax()]
print(f"b) Mês de maior faturamento: {mes_recorde['mes']} com faturamento de R$ {mes_recorde['faturamento_rs']:,.2f}")

# c) Faturamento total obtido com a comercialização do Produto A
faturamento_total = df_vendas['faturamento_rs'].sum()
print(f'c) Faturamento total obtido: R$ {faturamento_total:,.2f}')

# d) Ano que apresentou a maior quantidade de unidades vendidas
df_vendas['ano'] = df_vendas['mes'].str.split('-').str[0]
vendas_ano = df_vendas.groupby('ano')['unidades_vendidas'].sum()
print(f'\nd) Vendas anuais acumuladas:\n{vendas_ano}')
print(f"   -> Ano com maior volume de vendas: {vendas_ano.idxmax()} ({vendas_ano.max():,} unidades vendidas)")

# e) Relação entre preço unitário e unidades vendidas
corr_preco_vendas = df_vendas[['preco_unitario_rs', 'unidades_vendidas']].corr().loc['preco_unitario_rs', 'unidades_vendidas']
print(f'\ne) Correlação entre Preço Unitário e Unidades Vendidas: r = {corr_preco_vendas:.4f}')
print('   -> Há forte correlação positiva (0,77): ao longo dos anos, o volume de unidades vendidas continuou crescendo junto com os reajustes no preço unitário do produto.')
```
