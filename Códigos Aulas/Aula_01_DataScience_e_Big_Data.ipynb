```python
import pandas as pd

dados = {
    "produto": ["Motor A", "Motor B", "Motor A", "Motor C", "Motor B"],
    "producao": [100, 120, 110, 90, 130],
    "defeitos": [5, 12, 6, 3, 15],
    "temperatura": [70, 82, 73, 68, 85]
}

df = pd.DataFrame(dados)
```

```python
import pandas as pd

# Criar o DataFrame
df = pd.DataFrame({
    "maquina": ["A", "B", "C", "D", "E"],
    "producao": [1250, 1380, 1190, 1450, 1320],
    "energia": [420, 465, 390, 510, 405]
})

# a) Máquina que mais produziu
print(df.loc[df["producao"].idxmax()])

# b) Máquina que menos consumiu energia
print(df.loc[df["energia"].idxmin()])

# c) Produção média
print(df["producao"].mean())

# d) Consumo médio
print(df["energia"].mean())

# e) Ordenar pela produção
print(df.sort_values("producao", ascending=False))

# f) Calcular eficiência
df["eficiencia"] = df["producao"] / df["energia"]

# g) Máquina mais eficiente
print(df.loc[df["eficiencia"].idxmax()])

# Exibir DataFrame completo
print(df)
```

```python
import pandas as pd

# Criar o DataFrame
df = pd.DataFrame({
    "maquina": ["A", "B", "C", "D", "E"],
    "producao": [1250, 1380, 1190, 1450, 1320],
    "energia": [420, 465, 390, 510, 405]
})

# a) Máquina que mais produziu
print("Maquina que mais produziu:",df.loc[df["producao"].idxmax()])

# b) Máquina que menos consumiu energia
print("Maquina que menos produziu:",df.loc[df["energia"].idxmin()])

# c) Produção média
print("Produção média:",df["producao"].mean())

# d) Consumo médio
print("Consumo médio:",df["energia"].mean())

# e) Ordenar pela produção
print(df.sort_values("producao", ascending=False))

# f) Calcular eficiência
df["eficiencia"] = df["producao"] / df["energia"]
print("Eficiência: ",df['eficiencia'])

# g) Máquina mais eficiente
print("Maquina mais eficiente:", df.loc[df["eficiencia"].idxmax()])

# Exibir DataFrame completo
print(df)
```
